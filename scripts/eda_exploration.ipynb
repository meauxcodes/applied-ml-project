```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

clean_data = pd.read_csv("../data/clean/cleaned_garmin_data.csv")


corr_pairs = clean_data.corr(numeric_only=True).unstack()
corr_pairs = corr_pairs[corr_pairs.index.get_level_values(0) != corr_pairs.index.get_level_values(1)]
corr_pairs_sorted = corr_pairs.reindex(corr_pairs.abs().sort_values(ascending=False).index).to_string()
print(corr_pairs_sorted)
```
